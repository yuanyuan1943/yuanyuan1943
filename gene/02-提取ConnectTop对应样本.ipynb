```python
import os
import pandas as pd
from pathlib import Path
```

```python
gene_df = pd.read_csv(valid_gene_file_path, index_col=0)
```

```python
module_color2sample_dict = {}
for root, _, files in os.walk(data_output_dir_path / "connect_top"):
    for file in files:
        if file.endswith("_connect_top.csv"):
            # print(os.path.join(root, file))
            df = pd.read_csv(os.path.join(root, file))
            file = file.replace("_connect_top.csv", "")
            if file not in module_color2sample_dict:
                module_color2sample_dict[file] = df['sample'].to_list()
            else:
                module_color2sample_dict[file] = module_color2sample_dict[file].append(df['sample'].to_list())
```

```python
color2count = [(color, len(sample_key_list)) for color, sample_key_list in module_color2sample_dict.items()]
color2count = sorted(color2count, key=lambda x: x[1], reverse=True)
```

```python
sample_key_list = []
for color, _ in color2count:
    key_list = module_color2sample_dict[color]
    sample_key_list.extend(key_list)
```

```python
wgcna_df = gene_df.loc[sample_key_list]
wgcna_df.to_csv(data_output_dir_path / "wgcna.csv")
print(wgcna_df.shape)
```

```python
copy_number_df = pd.read_csv(copy_number_file_path, sep="\t", index_col=0)
copy_number_df = pd.read_csv(copy_number_thresholded_file_path, sep="\t", index_col=0)
```

```python
copy_number_key_list = copy_number_df.index.tolist()
not_in_copy_number_key_list = []
for key in sample_key_list:
    if key not in copy_number_key_list:
        not_in_copy_number_key_list.append(key)
print(f"基因样本数量: {len(sample_key_list)}")
print("拷贝数缺失数量: ", len(not_in_copy_number_key_list), not_in_copy_number_key_list[:5])

for key in not_in_copy_number_key_list:
    new_row = {col:0 for col in copy_number_df.columns}
    copy_number_df.loc[key] = new_row
copy_number_df = copy_number_df.loc[sample_key_list, :]
print(f"拷贝数样本数量: {len(copy_number_df)}")

copy_number_df.to_csv(data_output_dir_path / "copy-number.csv")
```
